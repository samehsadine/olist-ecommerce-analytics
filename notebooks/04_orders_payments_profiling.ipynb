## Charger les données


```python
import pandas as pd
from pathlib import Path

bronze_path = Path.cwd().parent / "data" / "bronze" / "parquet"
df_orders_payments = pd.read_parquet (bronze_path/"olist_order_payments_dataset.parquet")


print("Dimensions :", df_orders_payments.shape)

# Vérifier les types et valeurs non nulles
df_orders_payments.info()

df_orders_payments.head(5)

```

## Vérification des valeurs négatifs et à zéro

```python

# Vérifier les valeurs négatives
print("Montants négatifs :", (df_orders_payments["payment_value"] < 0).sum())

print("Versements négatifs :", (df_orders_payments["payment_installments"] < 0).sum())

print("Séquences négatives :", (df_orders_payments["payment_sequential"] < 0).sum())

# Vérifier les valeurs à zéro
print("Montants à zéro :", (df_orders_payments["payment_value"] == 0).sum())

print("Versements à zéro :", (df_orders_payments["payment_installments"] == 0).sum())

print("Séquences à zéro :", (df_orders_payments["payment_sequential"] == 0).sum())
```

## Examiner les lignes suspectes

```python
# Afficher les paiements dont le montant est égal à zéro
df_orders_payments[df_orders_payments["payment_value"] == 0]
```

```python
#versement à 0`
df_orders_payments[
    df_orders_payments["payment_installments"] == 0
]
```

## Vérification des doublons 

```python
# Vérifier leslignes entièrement dupliquées

print (df_orders_payments.duplicated().sum())

# Vérifier les doublons sur la clé composée
print("Clé(order_id, payment_sequential ) dupliuée :",
       df_orders_payments.duplicated(
        subset=["order_id", "payment_sequential"]
    ).sum()
      
    )
```
