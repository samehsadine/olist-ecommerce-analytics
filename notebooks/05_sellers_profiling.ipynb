## Charger les données


```python
import pandas as pd
from pathlib import Path

bronze_path = Path.cwd().parent / "data" / "bronze" / "parquet"
df_sellers = pd.read_parquet (bronze_path/"olist_sellers_dataset.parquet")

print("Dimensions :", df_sellers.shape)

# Vérifier les types et valeurs non nulles
df_sellers.info()

df_sellers.isna().sum()

df_sellers.head(5)
```

```python
df_sellers.isna().sum()
```

## Vérification des doublons 

```python
print (df_sellers.duplicated().sum())

# Vérifier les doublons sur la clé seller_id
print("Clé(seller_id ) dupliuée :",
       df_sellers.duplicated(
        "seller_id"
    ).sum()
      
    )
```

## Vérification les codes postaux

```python
zip_codes = (
    df_sellers["seller_zip_code_prefix"]
    .astype("string")
    .str.zfill(5)
)

zip_invalides = ~zip_codes.str.fullmatch(r"\d{5}")

print("Codes postaux invalides :", zip_invalides.sum())
```

## Vérification les villes et les États

```python
# Vérifier les villes vides
print(
    "Villes vides :",
    df_sellers["seller_city"].str.strip().eq("").sum()
)

# Vérifier les États vides
print(
    "États vides :",
    df_sellers["seller_state"].str.strip().eq("").sum()
)

# Vérifier le nombre d'États différents
print(
    "Nombre d'États :",
    df_sellers["seller_state"].nunique()
)
```

## Valider les codes des États

```python
# Liste des codes officiels des États brésiliens
etats_valides = {
    "AC", "AL", "AP", "AM", "BA", "CE", "DF",
    "ES", "GO", "MA", "MT", "MS", "MG", "PA",
    "PB", "PR", "PE", "PI", "RJ", "RN", "RS",
    "RO", "RR", "SC", "SP", "SE", "TO"
}

# Vérifier les États qui ne figurent pas dans la liste
etats_invalides = ~df_sellers["seller_state"].isin(etats_valides)

print("États invalides :", etats_invalides.sum())
```
