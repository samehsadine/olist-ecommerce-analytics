# Product Category Translation Data Profiling


## Load the Data

```python
import pandas as pd
from pathlib import Path

bronze_path = Path.cwd().parent / "data" / "bronze" / "parquet"
df_cat_translation_name = pd.read_parquet (bronze_path/"product_category_name_translation.parquet")

print("Dimensions :", df_cat_translation_name.shape)

# Check data types and non-null values
df_cat_translation_name.info()

df_cat_translation_name.isna().sum()

df_cat_translation_name.head(5)
```

```python
# Check fully duplicated rows
print(
    "Lignes dupliquées :",
    df_cat_translation_name.duplicated().sum()
)

# Check duplicate Portuguese categories
print(
    "Catégories dupliquées :",
    df_cat_translation_name["product_category_name"].duplicated().sum()
)

# Check the number of unique categories
print(
    "Catégories uniques :",
    df_cat_translation_name["product_category_name"].nunique()
)
```

## Check Categories Without Translations

```python
# Load the Products dataset
df_products = pd.read_parquet(
    bronze_path / "olist_products_dataset.parquet"
)

# Retrieve product categories
categories_products = set(
    df_products["product_category_name"].dropna()
)

# Retrieve categories available in the translation file
translation_categories = set(
    df_cat_translation_name["product_category_name"].dropna()
)

# Find categories without translations
categories_without_translation = (
    categories_products - translation_categories
)

print("Nombre de catégories sans traduction :", len(categories_without_translation))
print("Catégories sans traduction :", categories_without_translation)
```

## Check Missing Values

```python
# Check NULL (NaN) values
print("Valeurs manquantes :")
print(df_cat_translation_name.isna().sum())

# Check empty strings or strings containing only whitespace
for column in df_cat_translation_name.columns:

    empty_values = (
        df_cat_translation_name[column]
        .fillna("")
        .str.strip()
        .eq("")
    )

    print(
        f"{column} : {empty_values.sum()} valeurs vides ou manquantes"
    )
```
