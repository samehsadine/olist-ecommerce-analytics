# Products Data Profiling


## Load the Data

```python
import pandas as pd
from pathlib import Path

bronze_path = Path.cwd().parent / "data" / "bronze" / "parquet"
df_products = pd.read_parquet (bronze_path/"olist_products_dataset.parquet")

print("Dimensions :", df_products.shape)

# Check data types and non-null values
df_products.info()

df_products.isna().sum()

df_products.head(5)
```

## Check Duplicates

```python
print("Identifiants de produits uniques :", df_products["product_id"].nunique())
print("Nombre total de produits :", len(df_products))
```

## Check Numeric Values

```python
# Numeric columns to check
numeric_columns = [
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

# Check negative and zero values
for column in numeric_columns:
    print(f"{column} :")
    print("Valeurs négatives :", (df_products[column] < 0).sum())
    print("Valeurs à zéro :", (df_products[column] == 0).sum())
    print("----------------------")
```

```python
# Display products with zero weight
df_products[df_products["product_weight_g"] == 0]
```

## Check Categories

```python
# Number of distinct categories
print(
    "Nombre de catégories :",
    df_products["product_category_name"].nunique()
)

# Display the most frequent categories
print(
    df_products["product_category_name"].value_counts().head(10)
)
```

## Check Empty Categories

```python
# Check empty categories
empty_categories = (
    df_products["product_category_name"]
    .fillna("")
    .str.strip()
    .eq("")
)

print("Catégories vides ou manquantes :", empty_categories.sum())
```

## Check Maximum Text Lengths


```python
# Check the maximum ID length
print(
    "Longueur maximale product_id :",
    df_products["product_id"].str.len().max()
)

# Check the maximum category length
print(
    "Longueur maximale catégorie :",
    df_products["product_category_name"].str.len().max()
)
```

## Numeric Value Ranges


```python
numeric_columns = [
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

print(df_products[numeric_columns].agg(["min", "max"]))
```
